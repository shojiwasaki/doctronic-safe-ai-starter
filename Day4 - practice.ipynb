```python
import matplotlib.pyplot as plt

success_by_dept = df.groupby('department')['ai_project_success'].mean().sort_values(ascending=False)

fig, ax = plt.subplots(figsize=(8, 5))
ax.bar(success_by_dept.index, success_by_dept.values, color='#3B82F6')
ax.set_ylabel('Success Rate')
ax.set_title('AI Project Success Rate by Department')
ax.set_ylim(0, 1)
plt.xticks(rotation=30, ha='right')
plt.tight_layout()
plt.show()
```

```python
import pandas as pd

df = pd.read_csv("AI_Business_Transformation_Synthetic_Dataset.csv")

print(df.shape)
df.head()
```

```python
df.info()
df.describe()
```

```python
df['ai_project_success'].value_counts()
df.groupby('department')['ai_project_success'].mean().sort_values(ascending=False)
import matplotlib.pyplot as plt

success_by_dept = df.groupby('department')['ai_project_success'].mean().sort_values(ascending=False)

fig, ax = plt.subplots(figsize=(8, 5))
ax.bar(success_by_dept.index, success_by_dept.values, color='#3B82F6')
ax.set_ylabel('Success Rate')
ax.set_title('AI Project Success Rate by Department')
ax.set_ylim(0, 1)
plt.xticks(rotation=30, ha='right')
plt.tight_layout()
plt.show()
```

```python
import seaborn as sns

numeric_cols = df.select_dtypes(include='number').columns

corr = df[numeric_cols].corr()

fig, ax = plt.subplots(figsize=(9, 7))
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm', center=0, ax=ax)
ax.set_title('Correlation Heatmap of Numeric Features')
plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(8, 5))
ax.scatter(df['process_volume_monthly'], df['ai_project_success'], alpha=0.4, color='#3B82F6')
ax.set_xlabel('Process Volume Monthly')
ax.set_ylabel('AI Project Success (0 = Fail, 1 = Success)')
ax.set_title('Process Volume vs AI Project Success')
plt.tight_layout()
plt.show()
```
